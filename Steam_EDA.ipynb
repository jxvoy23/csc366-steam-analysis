Due to how some of the data in the csv is displayed awkwardly, we need to first check whether the columns have viable data before doing any analysis.

```python
df_eda['genres'].head()
```

As an example shown above, we need to fix how the data for the genres is displayed for proper data analysis.

```python
#Transform genres into a readable format
import ast

def clean_genres(genre_string):
    if pd.isna(genre_string):
        return "Unknown"

    try:
        genres = ast.literal_eval(genre_string)
        return ", ".join(genre["description"] for genre in genres)
    except (ValueError, TypeError):
        return "Unknown"

df_eda["genres"] = df_eda["genres"].apply(clean_genres)
df_eda["genres"]
```
