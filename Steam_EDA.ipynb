Due to how some of the data in the csv is displayed awkwardly, we need to first check whether the columns have viable data before doing any analysis.

```python
df_eda['genres'].head()
```

As an example shown above, we need to fix how the data for the genres is displayed for proper data analysis.

```python
#Transform genres into a readable format
import ast

def clean_genres(genre_string):
    if pd.isna(genre_string):
        return "Unknown"

    try:
        genres = ast.literal_eval(genre_string)
        return ", ".join(genre["description"] for genre in genres)
    except (ValueError, TypeError):
        return "Unknown"

df_eda["genres"] = df_eda["genres"].apply(clean_genres)
df_eda["genres"]
```

All of the cells below show some sort of data cleaning and analysis for the review ratings that the overall steam dataset has. It allows us to see how well a percentage of games are doing as well as checking the positive review percentage for games to see their ratings.

```python
# Checking review rating categories

df_eda["review_rating"].value_counts(dropna=False)
```

```python
#Calculating positive review percentage

df_eda["positive_review_pct"] = (
    df_eda["positive_reviews"] /
    (df_eda["positive_reviews"] + df_eda["negative_reviews"])
) * 100
```

From cleaning the data we can see that there are minimal missing data points. However they do consist of not having numerical numbers which is not bad since we can conclude they are newer games or yet to release.

```python
# Compare SteamSpy and Steam Store prices for USD games

usd_price_check = df_eda[
    (df_eda["currency"] == "USD") &
    (df_eda["store_price"].notna()) &
    (df_eda["steamspy_price"].notna())
].copy()

print("Number of USD games with both prices:", len(usd_price_check))
```

```python
# Calculate the difference between the two price sources

usd_price_check["price_difference"] = (
    usd_price_check["steamspy_price"] - usd_price_check["store_price"]
).round(2)

usd_price_check["price_match"] = (
    usd_price_check["price_difference"].abs() <= 0.01
)

print("Matching prices:", usd_price_check["price_match"].sum())
print("Different prices:", (~usd_price_check["price_match"]).sum())
print("Percent matching:", round(usd_price_check["price_match"].mean() * 100, 2), "%")
```

```python
# Display games with the largest price differences

usd_price_check.reindex(
    usd_price_check["price_difference"].abs().sort_values(ascending=False).index
)[["game_name", "steamspy_price", "store_price", "price_difference"]].head(20)
```

## Below will be the EDA for the dataset

```python
# Price summary statistics
print("Price Summary:")
print(df_eda["steamspy_price"].describe())

# Free vs. paid game counts
print("\nFree vs. Paid Games:")
print(df_eda["is_free"].value_counts(dropna=False))

# Number of games with $0 SteamSpy price
print("\nGames with $0 price:", (df_eda["steamspy_price"] == 0).sum())
```

```python
# Rating summary statistics
print("Positive Review Percentage Summary:")
print(df_eda["positive_review_pct"].describe())

# Review rating categories
print("\nReview Rating Distribution:")
print(df_eda["review_rating"].value_counts())

# Total review counts summary
print("\nTotal Reviews Summary:")
print(df_eda["total_reviews"].describe())
```

```python
# Price vs. positive review percentage
price_rating_df = df_eda[
    ["steamspy_price", "positive_review_pct"]
].dropna()

# Pearson correlation
pearson_corr = price_rating_df["steamspy_price"].corr(
    price_rating_df["positive_review_pct"]
)

# Spearman correlation
spearman_corr = price_rating_df["steamspy_price"].corr(
    price_rating_df["positive_review_pct"], method="spearman"
)

print("Games analyzed:", len(price_rating_df))
print("Pearson correlation:", round(pearson_corr, 3))
print("Spearman correlation:", round(spearman_corr, 3))
```
