The data does not show a lot of missing data except in the columns price overview final and price overview currency which might result in missing values due to pricing differences. Or different pricing in regions. We also can not forget that the dataset will include a lot of different currencies so we need to have one consistent one.

```python
missing_values = df_eda.isnull().sum()

missing_percentage = (df_eda.isnull().sum() / len(df_eda)) * 100

missing_summary = pd.DataFrame({
    "Missing Count": missing_values,
    "Missing Percentage": missing_percentage
})

missing_summary.sort_values(
    by="Missing Count",
    ascending=False
)
```

Due to how some of the data in the csv is displayed awkwardly, we need to first check whether the columns have viable data before doing any analysis.

```python
df_eda['store_data.genres'].head()
```

As an example shown above, we need to fix how the data for the genres is displayed for proper data analysis.

```python
#Transform genres into a readable format
import ast

def clean_genres(genre_string):
    if pd.isna(genre_string):
        return "Unknown"

    try:
        genres = ast.literal_eval(genre_string)
        return ", ".join(genre["description"] for genre in genres)
    except (ValueError, TypeError):
        return "Unknown"

df["genres_clean"] = df["store_data.genres"].apply(clean_genres)
df["genres_clean"]
```
